```python
import sys
sys.path.insert(0, 'D:/CADD/workspace/qclaw/projects/face-photo-sort')
import sqlite3
from pathlib import Path

BASE = Path('D:/CADD/workspace/qclaw/projects/face-photo-sort')
DB = BASE / 'databases' / 'photo_index.db'

conn = sqlite3.connect(str(DB))
conn.row_factory = sqlite3.Row
cur = conn.execute('PRAGMA table_info(photos)')
cols = [r['name'] for r in cur.fetchall()]
print('Columns:', cols)
conn.close()
```

```python
conn = sqlite3.connect(str(DB))
conn.row_factory = sqlite3.Row

# 先抽样 5 条，看 raw 数据
rows = conn.execute('SELECT * FROM photos LIMIT 5').fetchall()
for r in rows:
    print(dict(r))
conn.close()
```

```python
conn = sqlite3.connect(str(DB))
conn.row_factory = sqlite3.Row

# 文件名和 md5 字段
cur = conn.execute('SELECT path, md5 FROM photos LIMIT 20')
rows = cur.fetchall()
print(f'Total rows in DB: {conn.execute("SELECT COUNT(*) FROM photos").fetchone()[0]}')
print()
for r in rows:
    print(f'path : {r["path"]}')
    print(f'md5  : {r["md5"]}')
    print()
conn.close()
```

```python
# 有 md5 的 vs 没 md5 的
conn = sqlite3.connect(str(DB))
has_md5 = conn.execute('SELECT COUNT(*) FROM photos WHERE md5 IS NOT NULL AND md5 != ""').fetchone()[0]
no_md5  = conn.execute('SELECT COUNT(*) FROM photos WHERE md5 IS NULL OR md5 == ""').fetchone()[0]
total   = conn.execute('SELECT COUNT(*) FROM photos').fetchone()[0]
print(f'Total  : {total}')
print(f'Has md5: {has_md5}')
print(f'No md5 : {no_md5}')
conn.close()
```
